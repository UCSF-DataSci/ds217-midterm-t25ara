```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Import utilities from Q3
from q3_data_utils import load_data, detect_missing, fill_missing

# Load the data
df = load_data('data/clinical_trial_raw.csv')
print(f"Loaded {len(df)} patients")

# Prewritten visualization function for missing data
def visualize_missing_data(missing_counts):
    """
    Create a bar chart of missing values by column.
    
    Args:
        missing_counts: pandas Series with missing value counts per column
    """
    plt.figure(figsize=(10, 6))
    missing_counts.plot(kind='bar')
    plt.title('Missing Values by Column')
    plt.xticks(rotation=45)
    plt.ylabel('Number of Missing Values')
    plt.tight_layout()
    plt.show()
```

## Part 1: Detect Missing Data (3 points)

1. Use the `detect_missing()` utility to find missing values
2. Visualize missing data with a bar plot
3. Calculate the percentage of missing values per column

```python
# TODO: Detect and analyze missing data
# 1. Use detect_missing(df) to get missing value counts
# 2. Calculate percentage of missing values per column  
# 3. Print both counts and percentages
# 4. Identify which columns have missing data

# Optional: Use the visualization function above to create a bar chart
# visualize_missing_data(missing_counts)
def detect_missing(df: pd.DataFrame) -> pd.Series:
    """
    Return count of missing values per column.

    Args:
        df: Input DataFrame

    Returns:
        pd.Series: Count of missing values for each column

    Example:
        >>> missing = detect_missing(df)
        >>> missing['age']
        15
    """
    missing_val = df.isnull().sum()
    return missing_val
miss = detect_missing(df)
print("Number of Missing Entries Per Column:")
display(miss)
miss_percent = (miss/len(df))*100
print("Percent Missing Data Per Column:")
display(miss_percent)

missing_summary = miss[miss>0]
print(f"Columns with missing data are:\n {missing_summary}")
```

## Part 2: Compare Imputation Strategies (6 points)

For the 'cholesterol_total' column (which has missing values):

1. Fill with mean using `fill_missing()` utility
2. Fill with median using `fill_missing()` utility  
3. Forward fill using pandas `.fillna(method='ffill')`
4. Compare the three strategies - create a summary table showing:
   - Original mean/median
   - Mean/median after each strategy
   - How many values were filled

```python
# TODO: Compare imputation strategies
#display(df['cholesterol_total'].isnull().head(20))
#number 15 is a missing value
df_filled = fill_missing(df, 'cholesterol_total', strategy='mean')
meanfilled = (df_filled.loc[15,'cholesterol_total'])
print(f'mean filled missing value: {meanfilled}')
df_filledmed = fill_missing(df, 'cholesterol_total', strategy='median')
medfilled = (df_filledmed.loc[15,'cholesterol_total'])
print(f'med filled missing value: {medfilled}')
df_ffilled = fill_missing(df, 'cholesterol_total', strategy='ffill')
ffilled = (df_ffilled.loc[15,'cholesterol_total'])
print(f'f filled missing value: {ffilled}')
original_mean = (df['cholesterol_total'].mean())
meanfilled_mean = (df_filled['cholesterol_total'].mean())
medianfilled_mean = (df_filledmed['cholesterol_total'].mean())
ffill_mean = (df_ffilled['cholesterol_total'].mean())
original_med = (df['cholesterol_total'].median())
meanfilled_med = (df_filled['cholesterol_total'].median())
medianfilled_med = (df_filledmed['cholesterol_total'].median())
ffill_med = (df_ffilled['cholesterol_total'].median())
values_filled = 554
summary = pd.DataFrame({
    'Strategy': ['Original', 'Mean Fill', 'Median Fill', 'Forward Fill'],
    'Mean': [original_mean, meanfilled_mean, medianfilled_mean, ffill_mean],
    'Median': [original_med, meanfilled_med, medianfilled_med, ffill_med],
    'Values Filled': [0, values_filled, values_filled, values_filled]  # 0 for original, 554 for others
})
display(summary)
```
